## 4. Discover and validate the recording manifest
This expects each WAV and RTTM to have the same basename. Use only 2–3 short clips for the pilot.

````python
import pandas as pd, soundfile as sf, hashlib
wavs = sorted(DATA_DIR.rglob("*.wav"))
rows = []
for wav in wavs:
    rttm = wav.with_suffix(".rttm")
    if not rttm.exists():
        print("Skipping (no matching RTTM):", wav)
        continue
    info = sf.info(str(wav))
    rows.append({"id":wav.stem, "audio":str(wav), "reference_rttm":str(rttm),
                 "duration_s":round(info.duration,2), "sample_rate":info.samplerate})
manifest = pd.DataFrame(rows)
assert 2 <= len(manifest) <= 3, f"Expected 2–3 WAV/RTTM pairs; found {len(manifest)}"
assert manifest.id.is_unique, "Clip IDs must be unique"
display(manifest)
manifest.to_csv("/content/manifest.csv",index=False)
````

## 5. Parse RTTM and define the exact scoring region
Speaker IDs are arbitrary labels; the scorer optimally maps predicted IDs to reference IDs. The default scored region is the whole recording.

````python
from pyannote.core import Annotation, Segment, Timeline
from pyannote.metrics.diarization import DiarizationErrorRate, JaccardErrorRate

def read_rttm(path, clip_id):
    ann = Annotation(uri=clip_id)
    with open(path, encoding="utf-8") as f:
        for line in f:
            if not line.strip() or line.startswith("#"):
                continue
            x = line.split()
            if x[0] != "SPEAKER":
                continue
            if x[1] != clip_id:
                # Some public datasets use a different recording ID; this is okay
                # when the RTTM was verified to match this exact WAV.
                pass
            start, dur, speaker = float(x[3]), float(x[4]), x[7]
            if dur > 0:
                ann[Segment(start, start + dur), speaker] = speaker
    return ann

def write_rttm(ann, path, clip_id):
    with open(path, "w", encoding="utf-8") as f:
        for segment, _, label in ann.itertracks(yield_label=True):
            f.write(f"SPEAKER {clip_id} 1 {segment.start:.3f} {segment.duration:.3f} <NA> <NA> {label} <NA> <NA>\n")

def as_annotation(output):
    if isinstance(output, Annotation):
        return output
    if hasattr(output, "speaker_diarization"):
        return output.speaker_diarization
    raise TypeError(f"Unknown pipeline output type: {type(output)}")

def sha256(path):
    h=hashlib.sha256()
    with open(path,"rb") as f:
        for chunk in iter(lambda:f.read(1024*1024),b""):
            h.update(chunk)
    return h.hexdigest()

for r in manifest.itertuples():
    ann=read_rttm(r.reference_rttm,r.id)
    assert len(ann)>0, f"No speaker annotations in {r.reference_rttm}"
    assert max(s.end for s in ann.get_timeline()) <= r.duration_s + 0.1, f"RTTM extends beyond audio for {r.id}"
print("Reference RTTMs parsed successfully.")
````

## 6. Load both diarization pipelines

Model APIs can change. This cell tries the modern `token=` signature first and falls back to `use_auth_token=` for older installations. If a model is gated, verify your Hugging Face account has accepted its terms. **No known speaker count is passed to either system.**

````python
import torch, time, platform, importlib.metadata
from pyannote.audio import Pipeline

MODELS = ["pyannote/speaker-diarization-3.1",
          "pyannote/speaker-diarization-community-1"]
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)
print("pyannote.audio:", importlib.metadata.version("pyannote.audio"))
print("pyannote.metrics:", importlib.metadata.version("pyannote.metrics"))

def load_pipeline(name):
    try:
        pipeline=Pipeline.from_pretrained(name, token=os.environ["HF_TOKEN"])
    except TypeError:
        pipeline=Pipeline.from_pretrained(name, use_auth_token=os.environ["HF_TOKEN"])
    if pipeline is None:
        raise RuntimeError(f"Unable to load {name}; check access and token")
    return pipeline.to(DEVICE)

pipelines={}
for name in MODELS:
    print("Loading:",name)
    pipelines[name]=load_pipeline(name)
print("Both pipelines loaded.")
````

## 7. Run inference and calculate DER/JER

**DER** is the total false-alarm, missed-speech, and speaker-confusion time divided by reference speaker-time. **JER** summarizes per-speaker overlap disagreement. We report each clip separately; the pilot is too small to establish a statistically reliable model winner.

`collar=0.0`, `skip_overlap=False`, and the full recording timeline are explicit. This notebook does not use oracle speaker counts or automatic threshold tuning.

````python
from pathlib import Path
import json
OUT=Path("/content/voice_arena_results")
OUT.mkdir(exist_ok=True)
results=[]
for name, pipeline in pipelines.items():
    model_slug=name.replace("/","__")
    for r in manifest.itertuples():
        print(f"Running {name} on {r.id}...")
        t0=time.perf_counter()
        output=pipeline(r.audio)
        runtime_s=time.perf_counter()-t0
        hypothesis=as_annotation(output)
        reference=read_rttm(r.reference_rttm,r.id)
        uem=Timeline([Segment(0.0,float(r.duration_s))], uri=r.id)
        der_metric=DiarizationErrorRate(collar=0.0,skip_overlap=False)
        jer_metric=JaccardErrorRate(collar=0.0,skip_overlap=False)
        der=float(der_metric(reference,hypothesis,uem=uem))
        jer=float(jer_metric(reference,hypothesis,uem=uem))
        prediction_file=OUT/f"{r.id}__{model_slug}.rttm"
        write_rttm(hypothesis,prediction_file,r.id)
        results.append({
            "clip_id":r.id, "model":name, "duration_s":r.duration_s,
            "der_pct":round(der*100,3), "jer_pct":round(jer*100,3),
            "reference_speakers":len(reference.labels()),
            "predicted_speakers":len(hypothesis.labels()),
            "speaker_count_abs_error":abs(len(reference.labels())-len(hypothesis.labels())),
            "inference_seconds":round(runtime_s,2),
            "audio_sha256":sha256(r.audio),
            "reference_sha256":sha256(r.reference_rttm),
            "prediction_rttm":str(prediction_file)
        })
        print(f"  DER={der*100:.2f}% | JER={jer*100:.2f}%")
df=pd.DataFrame(results)
display(df[["clip_id","model","der_pct","jer_pct","reference_speakers",
            "predicted_speakers","speaker_count_abs_error","inference_seconds"]])
df.to_csv(OUT/"results.csv",index=False)
with open(OUT/"results.json","w") as f:
    json.dump({"models":MODELS,"device":str(DEVICE),"pyannote_audio_version":importlib.metadata.version("pyannote.audio"),
               "pyannote_metrics_version":importlib.metadata.version("pyannote.metrics"),
               "scoring":{"collar_seconds":0.0,"skip_overlap":False,"speaker_count_oracle":False},
               "results":results},f,indent=2)
````

## 8. Compare errors visually

These are *actual computed results* only after the previous cell succeeds. A two- or three-recording pilot is for pipeline validation and qualitative error inspection, not publication-grade statistical significance.

````python
import matplotlib.pyplot as plt
ax=df.pivot(index="clip_id",columns="model",values="der_pct").plot.bar(figsize=(10,4),rot=0)
ax.set_ylabel("DER (%) — lower is better")
ax.set_title("Pilot diarization errors by recording")
plt.tight_layout()
plt.show()
display(df.groupby("model")[["der_pct","jer_pct","speaker_count_abs_error"]].mean().round(2))
````
